```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import fetch_california_housing

df = fetch_california_housing()
```

```python
df
```

```python
dataset = pd.DataFrame(df.data)
dataset
```

```python
dataset.columns=df.feature_names
dataset.head()
```

```python
x=dataset
y=df.target
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
```

```python
x_test = sc.transform(x_test)
x_test
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
```

```python
reg=LinearRegression()
reg.fit(x_train,y_train)
```

```python
mse=-cross_val_score(reg,x_train,y_train,scoring='neg_mean_squared_error',cv=5)
```

```python
np.mean(mse)
```

```python
reg_pred=reg.predict(x_test)
```

```python
reg_pred
```

```python
sns.displot(y_test-reg_pred)
```

```python
sns.displot(y_test-reg_pred,kind='kde')
```

```python
from sklearn.metrics import r2_score
```

```python
score=r2_score(y_test,reg_pred)
```

```python
score
```

# Ridge Regression

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
```

```python
rid_reg=Ridge()
rid_reg
```

```python
parameter ={'alpha':[1,2,5,10,20,30,40,50,60,80,90]}
ridcv=GridSearchCV(rid_reg,parameter,scoring='neg_mean_squared_error',cv=5)
ridcv.fit(x_train,y_train)
```

```python
print(ridcv.best_params_)
```

```python
print(ridcv.best_score_)
```

```python
rid_pred=ridcv.predict(x_test)
```

```python
sns.displot(rid_pred-y_test,kind='kde')
```

```python
score1=r2_score(y_test,rid_pred)
```

```python
score1
```

# Lasso Regression

```python
from sklearn.linear_model import Lasso
```

```python
las=Lasso()
las
```

```python
parameter ={'alpha':[1,2,5,10,20,30,40,50,60,80,90]}
lascv=GridSearchCV(las,parameter,scoring='neg_mean_squared_error',cv=5)
lascv.fit(x_train,y_train)
```

```python
print(ridcv.best_score_)
```
